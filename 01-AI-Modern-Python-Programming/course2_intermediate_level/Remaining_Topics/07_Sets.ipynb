## 1) `set.discard()` vs `set.remove()` — difference and examples

Short summary:
- `remove(x)` removes `x` from the set but raises `KeyError` if `x` is not present.
- `discard(x)` removes `x` if present and does nothing if `x` is absent (no error).

Use `discard` when you want to ensure an item is absent without handling exceptions.
Use `remove` when absence is unexpected and should be flagged.


```python
# Examples: discard() vs remove()
s = {1, 2, 3}
print("original:", s)

# discard: no error if element missing
s.discard(2)
print("after discard(2):", s)
s.discard(99)   # no error
print("after discard(99):", s)

# remove: raises KeyError if missing
s = {1, 3}
print("\nnew set for remove examples:", s)
s.remove(1)
print("after remove(1):", s)
try:
    s.remove(99)   # will raise KeyError
except KeyError as e:
    print("remove(99) raised KeyError as expected:", e)
```

## 2) Understanding rehashing and why a `set`'s iteration order can change

Key points:
- A Python `set` is a hash table under the hood. Each element is placed into buckets determined by its hash.
- When the set grows (or sometimes shrinks), Python may *rehash* the internal table (resize and redistribute elements) to maintain performance.
- Rehashing can change where elements live in the internal structure — hence *iteration order is not guaranteed* and can change across runs or after mutations.
- **Important teaching message:** `set` is an *unordered* collection. Do not rely on any particular order; use `sorted()` or convert to a list and sort if order matters.


```python
# Demonstration: iteration order can change after mutations (rehashing/resizing)
# Note: Python's exact rehashing/resizing heuristics may vary by version/implementation,
# but this demonstrates a common observation: order can change after adding/removing items.

s = set(range(10))
print("initial:", list(s))

# Add more items to potentially trigger resize/rehash behavior
for i in range(10, 200):
    s.add(i)
# iterate a few times to show a snapshot
print("after extending (one snapshot):", list(s)[:20])

# Now remove and re-add an element to show order can change
s.remove(5)
s.add(5)
print("after remove(5) and add(5):", list(s)[:20])

# Another approach: create a fresh set with same elements — order may differ
s2 = set(list(s))
print("fresh set with same elements (s2) snapshot:", list(s2)[:20])
```
