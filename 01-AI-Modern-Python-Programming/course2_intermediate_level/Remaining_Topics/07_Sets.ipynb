## 1) `set.discard()` vs `set.remove()` — difference and examples

Short summary:
- `remove(x)` removes `x` from the set but raises `KeyError` if `x` is not present.
- `discard(x)` removes `x` if present and does nothing if `x` is absent (no error).

Use `discard` when you want to ensure an item is absent without handling exceptions.
Use `remove` when absence is unexpected and should be flagged.


```python
# Examples: discard() vs remove()
s = {1, 2, 3}
print("original:", s)

# discard: no error if element missing
s.discard(2)
print("after discard(2):", s)
s.discard(99)   # no error
print("after discard(99):", s)

# remove: raises KeyError if missing
s = {1, 3}
print("\nnew set for remove examples:", s)
s.remove(1)
print("after remove(1):", s)
try:
    s.remove(99)   # will raise KeyError
except KeyError as e:
    print("remove(99) raised KeyError as expected:", e)
```
